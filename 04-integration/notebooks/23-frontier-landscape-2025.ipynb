## 3. Company Deep Dives

### 3.1 OpenAI — The Reasoning Pioneer

OpenAI's biggest paradigm shift since GPT-4 has been the **o1/o3 family**: models trained via reinforcement learning to produce explicit chains of thought before answering. This "reasoning via RL" approach represents a fundamentally different scaling axis — instead of making the model bigger, you let it think longer at inference time.

**GPT-5.4** (Mar 2026) introduced a 1M-token context window and dramatically improved autonomous workflow capabilities, self-reporting ~75% on OSWorld-Verified (Mar 2026; best reported results are now ~80%, vs ~72% for humans). It was superseded in Apr 2026 by **GPT-5.5 / GPT-5.5 Pro** (1M ctx; ~52% on FrontierMath Tiers 1-3 for 5.5 Pro, ~83% on Terminal-Bench 2.0 -- approximate). The **Agents SDK** has replaced the older Assistants API, reflecting OpenAI's bet that agents — not chatbots — are the future product form factor. The **Responses API** provides a unified streaming interface, and **Operator** is their consumer-facing agent product.

Strategically, OpenAI has become increasingly enterprise-focused, with massive revenue growth. They publish less research than they used to, which has drawn criticism from the research community. The o3 model demonstrated that test-time compute scaling could unlock capabilities (e.g., ARC-AGI) that pure pre-training scaling had not.

**Interview angle**: OpenAI interviews will test whether you understand the reasoning paradigm deeply — not just that o3 "thinks" but HOW the RL training works, what the reward signal is, and what the limitations are.

---

### 3.2 Anthropic — Safety as Strategy

Anthropic's core thesis is that frontier AI development and safety research must happen together, not separately. Their **Constitutional AI (CAI)** framework replaces human preference labels with principle-based self-critique, allowing models to be aligned without massive human annotation budgets.

On the **interpretability** front, Anthropic has pushed further than any other lab. Their work on **sparse autoencoders** and **feature visualization** (the "Scaling Monosemanticity" paper) showed that individual neurons are **polysemantic** (each fires for many unrelated concepts), and that sparse autoencoders can extract interpretable **features** as directions/combinations spanning many neurons — a breakthrough toward understanding what models actually know and believe.

The **Claude Agent SDK** and **MCP (Model Context Protocol)** represent Anthropic's agent ecosystem play. MCP is an open standard for connecting AI models to tools and data sources — think of it as "USB-C for AI." Their **"Building Effective Agents"** framework identifies 5 key patterns for reliable agent systems:
1. **Prompt chaining**: Break complex tasks into sequential steps
2. **Routing**: Classify inputs and route to specialized handlers
3. **Parallelization**: Run independent subtasks concurrently
4. **Orchestrator-workers**: A central model coordinates specialized sub-agents
5. **Evaluator-optimizer**: Generate then critique, iterating to improve

**Computer use** capabilities allow Claude to interact with GUIs directly, representing a step toward general-purpose digital agents.

**Interview angle**: Anthropic interviews emphasize alignment thinking. You need to articulate WHY safety matters, not just THAT it matters. Be ready to discuss Constitutional AI mechanics, interpretability approaches, and scalable oversight.

---

### 3.3 Google DeepMind — The Multimodal Giant

The merger of Google Brain and DeepMind created a research powerhouse with unique advantages: massive compute, proprietary data (Search, YouTube, Scholar), and deep expertise in both ML research and systems engineering.

**Gemini 3.1** (Feb 2026; GPQA-Diamond 94.3%, tied GPT-5.4 at 57 on the Artificial Analysis Intelligence Index at launch) is natively multimodal — trained from scratch on text, images, audio, and video together, not bolted on after the fact. **Flash-Lite** represents the efficiency frontier at $0.25/M input tokens ($1.50/M output), enabling deployment at Google scale. **Gemini 3.5 Flash** shipped in May 2026, and **3.5 Pro** is announced with a 2M-token context target.

**AlphaProof** applied RL to formal mathematics, proving IMO-level theorems — a landmark for AI reasoning. The **A2A (Agent-to-Agent)** protocol is Google's answer to MCP, enabling agents to communicate with each other rather than just with tools. **Project Mariner** (web agent) and **Project Astra** (multimodal assistant) represent their agent product bets.

**Interview angle**: DeepMind interviews are deeply technical. Expect questions on architecture decisions (why MoE? why native multimodal?), optimization at scale, and research methodology. They value mathematical rigor.

---

### 3.4 Meta — The Open-Source Powerhouse

Meta's Llama-era strategy was unique: give away the model weights and win on ecosystem. **Llama 4 Scout** (109B, 10M ctx) and **Llama 4 Maverick** (400B-A17B, 1M ctx), released Apr 2025, turned out to be Meta's LAST open-weight flagships: **Behemoth was never released**, and after the 2025 Superintelligence Labs reorg Meta shipped a **closed, API-only frontier model** (Apr 2026). The Llama 4 launch was also marred by an LMArena controversy: an experimental Maverick variant scored 1417, while the release version landed around ~1270.

Their **post-training recipe** is well-documented and instructive:
1. **SFT** (Supervised Fine-Tuning) on high-quality demonstrations
2. **Rejection Sampling** to generate multiple responses and keep the best
3. **DPO** (Direct Preference Optimization) to align with human preferences
4. **Iterate**: Repeat steps 2-3 with updated model as the generator

The **Purple Llama** safety suite provides open-source safety tools: CyberSecEval for cybersecurity, Llama Guard for content filtering, and Code Shield for code safety.

Meta publishes prolifically — their research output rivals academic institutions. Key papers span self-supervised learning (DINO, I-JEPA), language models, and safety.

**Interview angle**: Meta FAIR interviews value research depth and publication track record. Expect questions about why open-source matters, how to evaluate model safety at scale, and specific technical decisions in the Llama training pipeline.

### 3.5 DeepSeek — The Efficiency Disruptor

DeepSeek shocked the field when **R1** matched OpenAI's o1 on reasoning benchmarks at a fraction of the training cost, using a novel algorithm called **GRPO (Group Relative Policy Optimization)**.

GRPO's key insight: you don't need a separate critic/value network. Instead, sample a group of responses, compute their rewards, and normalize within the group. This eliminates the need for a reward model altogether when combined with **rule-based rewards** (e.g., "did the math answer match?", "is the code syntactically valid?").

**DeepSeek-V3** uses a 671B MoE architecture with innovative multi-head latent attention (MLA) for KV cache compression and auxiliary-loss-free load balancing; **V3.2** (Dec 2025, 671B-A37B) refined it, and the **V4 preview** (Apr 2026) spans **V4-Pro** (1.6T MoE) and **V4-Flash** (284B) with 1M context under an MIT license. Note: there was never an "R2" — reasoning was absorbed into the V-line, and **R1 was published in Nature (Sep 2025)**. **DeepSeek-Prover-V2** demonstrated strong formal mathematics capabilities.

Their open-weights strategy means anyone can study, fine-tune, and deploy their models. This has accelerated the entire field's understanding of reasoning model training.

**Interview angle**: DeepSeek's work is a goldmine for interview questions. Be ready to explain GRPO vs PPO vs DPO, why MoE architectures are efficient, and what rule-based rewards can and cannot capture.

## 4. Key Debates in the Field

These are the discussions that come up in interviews. You need to have a position on each.

### 4.1 "Are We Hitting Scaling Walls?"

**The claim**: Pre-training compute scaling (the "Chinchilla" approach of scaling data and parameters together) is showing diminishing returns. GPT-5 reportedly took significantly more compute than GPT-4 but the capability jump was smaller.

**The counter**: Scaling hasn't stopped — it's shifted. Post-training scaling (more RL, more RLHF data, more rejection sampling iterations) and test-time compute scaling (letting models think longer) are new frontiers with their own scaling laws. The o1/o3 results show dramatic improvements from test-time compute without changing the base model.

**Your position should consider**:
- Pre-training scaling may be hitting data walls (we're running out of high-quality internet text)
- Synthetic data could extend the runway but has its own limits (model collapse)
- Test-time compute is clearly working but may have different ceiling characteristics
- The economic question: even if scaling works, is it cost-effective?

---

### 4.2 "Open vs Closed Models?"

**The trend**: The gap between open and closed models narrowed dramatically through 2025 — DeepSeek's open V-line (now the MIT-licensed V4 preview), the open Qwen3.6 line, Mistral Large 3, and Kimi K2.6 are competitive on many benchmarks (Qwen3 famously handles 119 languages). But the frontier's very top remains closed, and Meta — the open-weights champion — pivoted to a closed API-only frontier model in Apr 2026.

**Arguments for open**:
- Accelerates research (everyone can build on the work)
- Enables fine-tuning for specific use cases
- Reduces concentration of power
- DeepSeek proved open models can match closed ones

**Arguments for closed**:
- Safety: harder to misuse what you can't download
- The LMArena top tier is fully proprietary (June 2026 top-3: Claude Opus 4.6/4.7 Thinking variants; Grok 4.1's Nov-2025 #1 lasted only days)
- Business model sustainability (Meta's 2026 closed pivot shows the pressure on the free-weights model)
- Alignment properties may not survive fine-tuning

---

### 4.3 "Safety Regulation?"

**Current landscape**:
- **EU AI Act**: Regulatory framework with risk-based classification. Frontier models face transparency and safety requirements.
- **US executive orders**: Reporting requirements for large training runs. Evolving policy landscape.
- **Voluntary commitments**: Frontier labs have made various safety pledges (RSPs, safety cases)

**The tension**: Regulation could slow US/EU development while unregulated labs (particularly in China) continue at full speed. But no regulation risks catastrophic misuse.

**Your position should consider**:
- What specific harms are we regulating against? (misuse? existential risk? bias?)
- Can regulation be technically informed enough to be effective?
- What's the right balance between innovation and caution?

---

### 4.4 "Is Test-Time Compute the New Frontier?"

**The evidence**: OpenAI's o1/o3 and DeepSeek's R1 demonstrate that letting models "think" longer at inference time dramatically improves performance on reasoning tasks. o3 solved ARC-AGI tasks that no amount of pre-training scaling had cracked.

**The mechanism**: Train models via RL to produce chains of thought. More compute at inference = longer chains = better reasoning. This creates a new scaling axis orthogonal to model size.

**Open questions**:
- Where are the diminishing returns? Does thinking 10x longer give 10x better answers?
- How to allocate compute optimally? (Snell et al. 2024: sometimes a small model thinking long beats a big model answering fast)
- Does this work for all tasks, or mainly math/code/logic?
- Can test-time compute be made efficient enough for real-time applications?

---

### 4.5 "Agents: Hype or Real?"

**Production deployments are emerging**: OpenAI's Operator for web tasks, Anthropic's Computer Use for GUI interaction, coding agents (Cursor, Claude Code, Devin) are seeing real usage.

**But reliability is still a concern**: Agent systems fail in ways that are hard to predict. Error compounds over multi-step trajectories. The gap between demo and production is large.

**The key insight**: The best agent systems today use simple patterns (prompt chaining, tool use) rather than complex autonomous loops. Anthropic's "Building Effective Agents" paper emphasizes this: start simple, add complexity only where needed.

---

### 4.6 "MCP vs A2A?"

**MCP (Model Context Protocol)** — Anthropic's open standard:
- Connects models to **tools and data sources**
- Analogous to USB-C: one standard protocol for all connections
- Growing ecosystem: 1000+ MCP servers, adopted by multiple IDEs and platforms

**A2A (Agent-to-Agent)** — Google's protocol:
- Connects **agents to other agents**
- Designed for multi-agent coordination
- Complementary to MCP rather than strictly competitive

**The reality**: These protocols solve different problems. MCP is about tool access; A2A is about agent coordination. The winner may be "both" — or a unified standard may emerge. What matters is that the industry is converging on the need for standardized agent communication.

## 5. What Interviewers Want to Hear

When an interviewer asks "What's happening in the field?" or "What are you excited about?", they're evaluating multiple things at once:

### 5.1 Awareness of the Landscape
- You know who's building what — not just the big names but the nuances
- You understand the key technical differences (not just "OpenAI is big and Anthropic is safe")
- You can name specific models, papers, and capabilities
- **Bad**: "I've been following OpenAI's work closely" (too vague, too narrow)
- **Good**: "The most interesting development recently has been the convergence around test-time compute scaling — OpenAI's o3 and DeepSeek's R1 independently arrived at RL-trained chain-of-thought reasoning, suggesting this is a robust paradigm rather than a one-off trick."

### 5.2 Opinions Backed by Reasoning
- You don't just know facts — you have a VIEW on what they mean
- Your opinions are nuanced, not tribal ("X lab is the best" is a red flag)
- You can argue for AND against a position
- **Bad**: "I think scaling laws will continue forever" (too confident, not nuanced)
- **Good**: "I think pre-training scaling is hitting diminishing returns, but post-training and test-time scaling are opening new frontiers. The key question is whether test-time scaling laws have similar or different ceiling characteristics."

### 5.3 Understanding of Trade-offs
- No single lab has "won" — each has real strengths AND weaknesses
- Technical decisions involve trade-offs, not right answers
- You can discuss why a lab made a specific choice and what they sacrificed

### 5.4 Research Taste
- What problems excite YOU and why — this reveals your identity as a researcher
- Interviewers want to see that you've thought deeply about what you'd work on
- The best answers connect your past work to the lab's open problems

### 5.5 Intellectual Honesty
- What we DON'T know is as important as what we do
- Admitting uncertainty on genuinely uncertain questions is a strength, not a weakness
- "I don't know, but here's how I'd think about it" is a great answer structure
- **Bad**: "RLHF is solved" (overconfident)
- **Good**: "RLHF works well in practice, but we don't fully understand why — the theoretical grounding for why preference-based learning generalizes so well is still an open question."

**Insider Tip:** Interviewers don't want you to recite model names and benchmark scores. They want to hear your ANALYSIS: 'The trend is toward...' 'I think the most important development is...' 'I disagree with X because...' Having a point of view is what separates senior candidates.

## 6. Forming Your Point of View

> Don't think about what sounds impressive. Think about what genuinely keeps you up at night. The best researchers are driven by curiosity, not career strategy.

#### Prompt 2: The Alignment Question
> **"What's the most important unsolved problem in AI alignment?"**
>
> This is especially important for Anthropic and OpenAI alignment team interviews. But even at other labs, showing you've thought about alignment demonstrates research maturity.
>
> Possible angles: scalable oversight, interpretability, robustness to distribution shift, value alignment across cultures, corrigibility, deceptive alignment.

#### Prompt 3: Your Prediction
> **"What's your prediction for the most impactful development in AI in the next 2 years?"**
>
> This reveals how you think about the trajectory of the field. Good answers are specific and actionable, not vague ("AI will get better").
>
> Example of a good prediction: "I think we'll see a shift from single-model scaling to multi-agent architectures where specialized models coordinate on complex tasks. The bottleneck will shift from model capability to reliable agent communication — which is why MCP and A2A matter."

#### Prompt 4: Your Contrarian View
> **"What do you think most people in the field are wrong about?"**
>
> This is where you show intellectual courage. Strong researchers have views that diverge from consensus — backed by evidence.
>
> Example: "I think the focus on benchmark performance is leading us astray. The models that score highest on LMArena aren't necessarily the most useful or the safest — they may just be optimized for the kinds of tasks that appear in benchmarks."

### How to Structure Your Answer

When asked about your research interests in an interview, use this structure:

1. **The problem**: "I'm most interested in X because..."
2. **Why it matters**: "This matters because if we solve it, we can..."
3. **Current state**: "The best current approaches are A and B, but they fail because..."
4. **Your angle**: "I think the key insight is... and I'd approach it by..."
5. **Concretely**: "The first experiment I'd run is..."

This structure shows that you can think from problem to solution, not just recite facts.

### Question 1 (OPINION): "Which frontier lab do you think is in the best position and why?"

**What the interviewer is testing:** Do you follow the industry? Can you reason about strategy, not just technology? Do you have your own perspective?

**Red flag answers:**
- "I don't really follow the industry" -- immediate concern about research engagement
- "They're all the same" -- shows lack of depth
- "Obviously [interviewer's lab]" -- transparent flattery, and interviewers hate it

**How to structure your answer:**

"I think [Lab X] has the strongest position right now because [specific technical/strategic reason]. But this isn't a winner-take-all situation -- [Lab Y] has a significant advantage in [specific area], and [Lab Z]'s approach to [specific thing] could be the right long-term bet. The key differentiator I'm watching is [specific technical or strategic factor]."

**Example answer (calibrate to the time of your interview):**

"I think Anthropic has the best technical position in alignment research -- their Constitutional AI work and interpretability breakthroughs give them a structural advantage in safety that others are still catching up to. But Google DeepMind has the strongest overall research organization and compute access, which matters for training frontier models. Meta's open-source strategy with Llama has created an enormous ecosystem advantage -- they may not have the most capable model, but they have the most deployed models. OpenAI has the strongest product-market fit and distribution. The factor I'm watching most closely is whether scaling compute or scaling inference-time reasoning produces more capability -- that could determine which lab's approach wins."

**Tailor to the lab you're interviewing at:**
- Interviewing at Meta FAIR? Emphasize the value of open research, the Llama ecosystem, the research freedom.
- Interviewing at Anthropic? Show genuine safety engagement, discuss alignment as a technical problem.
- Interviewing at Google DeepMind? Discuss the advantage of deep research plus product integration.
- Interviewing at OpenAI? Discuss the advantage of product feedback loops for research.

But always be honest. Don't trash competitors -- the industry is small and people move between labs.

### Question 2 (RESEARCH TASTE): "What's the most important open problem in AI right now?"

**What the interviewer is testing:** Research taste. Can you identify what matters, not just what's trending?

**Weak answer:** "Making models bigger" or "AGI" (too vague, no specific insight).

**Strong answer structure:**
1. Name a specific problem
2. Explain why it matters (not just technically, but for the field)
3. Describe promising approaches
4. Connect to your own experience

**Example answer:**

"I think the most important open problem is scalable oversight -- how do we provide reliable training signal for tasks that are too complex for humans to evaluate? Right now, RLHF works because humans can judge most model outputs. But as models tackle harder problems -- multi-step reasoning, long-horizon planning, scientific research -- human evaluation becomes unreliable. Process reward models are one promising direction: reward each step rather than just the final answer. Debate is another: have models argue both sides so humans can judge arguments rather than solutions. Evaluating even straightforward tasks reliably is already hard -- and the scalable oversight problem is orders of magnitude harder."

**Other strong problem picks:**
- **Reasoning generalization:** Models can solve in-distribution reasoning problems but struggle with novel problem types. Is this fundamental or fixable with better training?
- **Interpretability at scale:** We can understand small circuits but not full model behavior. Closing this gap is essential for alignment.
- **Multi-modal grounding:** Language models learn statistical patterns, not world models. How do we get genuine understanding?
- **Efficient training:** Current frontier models cost $100M+ to train. Making training 10-100x more efficient would democratize research.

The key: pick something you genuinely care about and can discuss for 10 minutes with follow-up questions.

---

**Q2: What is MCP (Model Context Protocol)?**

<details><summary>Answer</summary>
An open standard developed by Anthropic for connecting AI models to tools and data sources. Think of it as "USB-C for AI" — one protocol that works across different models and tool providers. It has been adopted by 1000+ tool servers and multiple platforms.
</details>

---

**Q3: What is the test-time compute paradigm?**

<details><summary>Answer</summary>
Instead of making models bigger (pre-training scaling), let models "think" longer at inference time. Models like o3 and R1 are trained via RL to produce explicit chains of thought. More compute at inference = longer reasoning chains = better answers, especially on math, code, and logic tasks.
</details>

---

**Q4: What is GRPO and how does it differ from PPO?**

<details><summary>Answer</summary>
GRPO (Group Relative Policy Optimization), developed by DeepSeek, eliminates the critic/value network needed in PPO. Instead, it samples a group of responses, computes their rewards, and normalizes within the group to get advantages. This reduces training cost significantly and can work with rule-based rewards (no reward model needed).
</details>

---

**Q5: What are the 5 agent patterns from Anthropic's "Building Effective Agents"?**

<details><summary>Answer</summary>
1. Prompt chaining (sequential steps), 2. Routing (classify and dispatch), 3. Parallelization (concurrent subtasks), 4. Orchestrator-workers (central coordinator + specialized sub-agents), 5. Evaluator-optimizer (generate then critique, iterate).
</details>

---

**Q6: What is Constitutional AI?**

<details><summary>Answer</summary>
Anthropic's alignment approach where instead of using human preference labels, the model critiques its own outputs against a set of written principles (a "constitution"). This enables alignment at scale without massive human annotation budgets. The model is first trained with SL on self-revised outputs, then with RL against a model-based harmlessness judge.
</details>

---

**Q7: How does Meta's post-training recipe work?**

<details><summary>Answer</summary>
Four iterative steps: (1) SFT on high-quality demonstrations, (2) Rejection sampling to generate multiple responses and keep the best, (3) DPO to align with human preferences, (4) Iterate — repeat steps 2-3 with the updated model as the generator.
</details>

---

**Q8: What is A2A and how does it differ from MCP?**

<details><summary>Answer</summary>
A2A (Agent-to-Agent) is Google's protocol for connecting agents to other agents, enabling multi-agent coordination. MCP connects models to tools and data sources. They're complementary: MCP handles tool access, A2A handles agent-to-agent communication.
</details>

---

**Q9: What is DeepSeek's V3 architecture?**

<details><summary>Answer</summary>
A 671B MoE (Mixture of Experts) model with innovations including multi-head latent attention (MLA) for KV cache compression and auxiliary-loss-free load balancing. It's designed for efficiency — only a fraction of parameters are active for each token. (Since superseded: V3.2 in Dec 2025 (671B-A37B), and the V4 preview in Apr 2026 — V4-Pro 1.6T MoE + V4-Flash 284B, 1M ctx, MIT license.)
</details>

---

**Q10: What is AlphaProof?**

<details><summary>Answer</summary>
Google DeepMind's system that applies RL to formal mathematics, successfully proving IMO (International Mathematical Olympiad) level theorems. It represents a milestone in AI reasoning, combining neural network pattern matching with formal verification.
</details>

---

**Q11: What is the current state of the "open vs closed" model debate?**

<details><summary>Answer</summary>
The gap narrowed dramatically through 2025: open lines like DeepSeek's MIT-licensed V4 preview, Qwen3.6, Mistral Large 3, and Kimi K2.6 are competitive on many benchmarks. However, the LMArena top tier (June 2026: Claude Opus 4.6/4.7 Thinking variants) is proprietary, and Meta pivoted to a closed API-only frontier model in Apr 2026. Key tensions: open models accelerate research and reduce power concentration, but closed models offer more control over safety and misuse.
</details>

---

**Q12: What are the main arguments that pre-training scaling is hitting diminishing returns?**

<details><summary>Answer</summary>
1. Data wall — we're exhausting high-quality internet text, 2. The jump from GPT-4 to GPT-5 required much more compute for a smaller capability gain, 3. Post-training and test-time compute are showing better ROI for capability improvement. Counter-argument: synthetic data could extend the runway, and new modalities (video, audio) are untapped data sources.
</details>

---

**Q13: What is Anthropic's interpretability research focused on?**

<details><summary>Answer</summary>
Sparse autoencoders to find interpretable features in neural networks, feature visualization to understand what models represent internally, and circuit analysis to trace how information flows through models. The "Scaling Monosemanticity" paper showed individual features corresponding to interpretable concepts. Goal: understand models well enough to verify alignment.
</details>

---

**Q14: What is the standard context window size at the frontier in June 2026?**

<details><summary>Answer</summary>
1 million tokens is now standard for frontier models (GPT-5.5, Claude (Opus 4.8 / Fable 5), Llama 4 Maverick, DeepSeek V4, and Grok 4.3 all support 1M; Gemini 3.5 Pro targets 2M; Llama 4 Scout reached 10M). This was 8K in early 2023 — a 125x increase in ~3 years.
</details>

---

**Q15: Why is MoE (Mixture of Experts) the dominant architecture trend?**

<details><summary>Answer</summary>
MoE allows models to have many parameters (capacity) while only activating a fraction for each token (efficiency). This means you get the knowledge of a 671B model at the inference cost of a much smaller model. Llama 4, DeepSeek V3/V4, the Qwen3 family, Mistral Large 3, and Kimi K2.6 all use MoE.
</details>

---

**Q16: What is Project Mariner?**

<details><summary>Answer</summary>
Google DeepMind's web agent project — an AI that can navigate and interact with websites to accomplish tasks on behalf of users. Part of their agent product strategy alongside Project Astra (multimodal assistant).
</details>

**Q17: What is Purple Llama?**

<details><summary>Answer</summary>
Meta's open-source AI safety suite, including: CyberSecEval (cybersecurity evaluation), Llama Guard (content safety classifier), and Code Shield (code safety analysis). It provides tools for evaluating and mitigating safety risks in LLM deployments.
</details>

**Q18: What is the EU AI Act?**

<details><summary>Answer</summary>
The European Union's comprehensive AI regulatory framework using risk-based classification. High-risk AI systems face stricter requirements including transparency, safety testing, and human oversight. Frontier foundation models face additional obligations around safety documentation and reporting.
</details>
